```python
import pandas as pd
from src.data_cleaning import IngestData
def ingesting(data_path:str) -> pd.DataFrame:
    """
    Ingest data from a CSV file.

    Args:
        data_path (str): Path to the CSV file.

    Returns:
        pd.DataFrame: DataFrame containing the ingested data.
    """
    # Initialize the IngestData class with the provided data path
    ingest_data = IngestData(data_path=data_path)
    df = ingest_data.get_data()
    
        
    
    # Return the DataFrame
    return df

```

```python
from zenml import step
import pandas as pd
from typing import NamedTuple
import logging
from src.data_cleaning import DataCleaning
from typing import Tuple
from ingest_data import ingesting

@step
def cleaning(DataFrame:pd.DataFrame) :
    """Clean the data and return train-test splits."""

    data = DataCleaning(DataFrame)
    data.preprocess()
    return data.x_train, data.x_test, data.y_train, data.y_test
    
    
if __name__ == "__main__":
    # Example usage
    data=ingesting("data/olist_orders_dataset.csv")
    cleaned_data = cleaning(data)
    print(cleaned_data)
```
